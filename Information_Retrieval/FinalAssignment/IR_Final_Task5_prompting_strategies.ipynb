```python
import os
import json
import tqdm
import sys
MAX_LEN = 512
```

## Load model

```python
from transformers import AutoTokenizer, LlamaForCausalLM, AutoModelForCausalLM

model_name = "HuggingFaceH4/zephyr-7b-beta"
tokenizer = AutoTokenizer.from_pretrained(model_name, truncation=True, padding=True, padding_side="left", maximum_length = 2048, model_max_length = 2048)
model = AutoModelForCausalLM.from_pretrained(model_name, load_in_4bit = True, device_map = 'auto')
tokenizer.pad_token = tokenizer.eos_token
model.generation_config.pad_token_id = model.generation_config.eos_token_id
```

## Query expansion

# **PRF**

```python
import gzip, csv
from datasets import load_dataset
from tqdm.auto import tqdm
import pytrec_eval
from statistics import mean
from sentence_transformers import CrossEncoder
MAX_LEN = 512

qrels = {}
with open('/content/2019qrels-pass.txt') as f:
    for line in f:
        qid, _, docid, rel = line.strip().split()
        qrels.setdefault(qid, {})[docid] = int(rel)
all_qids = set(qrels)

queries = {}
with gzip.open('/content/msmarco-test2019-queries.tsv.gz', 'rt', encoding='utf8') as f:
    reader = csv.reader(f, delimiter='\t')
    for qid, text in reader:
        if qid in all_qids:
            queries[qid] = text

candidates = {}
with gzip.open('/content/msmarco-passagetest2019-top1000.tsv.gz', 'rt', encoding='utf8') as f:
    reader = csv.reader(f, delimiter='\t')
    for row in reader:
        qid, docid = row[0], row[1]
        if qid in all_qids:
            candidates.setdefault(qid, []).append(docid)

# 4. Load passage-text mapping
passage_ds = load_dataset(
    'sentence-transformers/msmarco-corpus', 'passage', split='train'
)
passage_text = {str(pid): text for pid, text in zip(passage_ds['pid'], passage_ds['text'])}

device = next(model.parameters()).device
cross_encoder = CrossEncoder(
    'cross-encoder/ms-marco-MiniLM-L-6-v2', device=str(device)
)

run_orig = {}
for qid in tqdm(all_qids, desc='Scoring original runs'):
    docs = candidates.get(qid, [])
    if not docs:
        continue
    pairs = [(queries[qid], passage_text[d]) for d in docs]
    scores = cross_encoder.predict(pairs, batch_size=64).tolist()
    run_orig[qid] = {d: float(s) for d, s in zip(docs, scores)}


prf_expansions = {}
for qid, doc_scores in tqdm(run_orig.items(), desc='Generating PRF expansions'):
    # get top-3 doc IDs by original score
    top3 = sorted(doc_scores.items(), key=lambda x: x[1], reverse=True)[:3]
    passages = [passage_text[d] for d, _ in top3]

    raw_prompt = (
        "CoT/PRF\n"
        "Answer the following query based on the context:\n"
        f"Context: {passages[0]}\n"
        f"{passages[1]}\n"
        f"{passages[2]}\n"
        f"Query: {queries[qid]}\n"
        "Give the rationale before answering."
    )
    # tokenize + truncate
    inputs = tokenizer(
        raw_prompt,
        return_tensors="pt",
        truncation=True,
        max_length=MAX_LEN,
        padding="max_length",
    ).to(device)

    out = model.generate(**inputs, max_new_tokens=64)
    gen = tokenizer.decode(out[0], skip_special_tokens=True)
    prf_expansions[qid] = gen


run_prf = {}
for qid, expansion in tqdm(prf_expansions.items(), desc='Scoring PRF runs'):
    docs = candidates.get(qid, [])
    if not docs:
        continue
    exp_q = queries[qid] + ' ' + expansion
    pairs = [(exp_q, passage_text[d]) for d in docs]
    scores = cross_encoder.predict(pairs, batch_size=64).tolist()
    run_prf[qid] = {d: float(s) for d, s in zip(docs, scores)}

# 9. Filter common qids and evaluate
common_qids = set(run_orig) & set(run_prf) & all_qids
if not common_qids:
    raise ValueError(f"No common qids to evaluate: {common_qids}")
evaluator = pytrec_eval.RelevanceEvaluator(
    {qid: qrels[qid] for qid in common_qids},
    {'ndcg_cut.10', 'recall.100', 'map_cut.1000'}
)
metrics_orig = evaluator.evaluate({qid: run_orig[qid] for qid in common_qids})
metrics_prf  = evaluator.evaluate({qid: run_prf [qid] for qid in common_qids})


def summarize(metrics):

    return {
        'NDCG@10': mean(m['ndcg_cut_10'] for m in metrics.values()),
        'Recall@100': mean(m['recall_100'] for m in metrics.values()),
        'MAP@1000': mean(m['map_cut_1000'] for m in metrics.values()),
    }
print('Before expansion:', summarize(metrics_orig))
print('PRF expansion:',   summarize(metrics_prf))
```

## Bonus Prompt Strategies


```python

prompt_templates = {
    'CoT': (
        "Answer the following query:\n{query}\nGive the rationale before answering."
    ),
    'CoT_distilled': (
        "Reformulate the query concisely.\nQuery: {query}\nAnswer:"  # outputs short reformulation
    ),
    'Q2E': (
        "Write a list of keywords for the given query:\nQuery: {query}\nKeywords:"
    ),
    'Q2E/PRF': (
        "Write a list of keywords based on context:\nContext:\n{docs}\nQuery: {query}\nKeywords:"
    )
}


device = next(model.parameters()).device
cross_encoder = CrossEncoder('cross-encoder/ms-marco-MiniLM-L-6-v2', device=str(device))


def score_run(run_qrels, run):
    return {qid: run[qid] for qid in run}

baseline_run = {}
for qid in tqdm(all_qids, desc='Scoring baseline'):
    docs = candidates.get(qid, [])
    if not docs: continue
    pairs = [(queries[qid], passage_text[d]) for d in docs]
    scores = cross_encoder.predict(pairs, batch_size=64).tolist()
    baseline_run[qid] = {d: float(s) for d, s in zip(docs, scores)}

# Iterate over prompt strategies
results = {}
for name, tmpl in prompt_templates.items():
    expansions = {}
    # generate expansions
    for qid in tqdm(all_qids, desc=f'Expanding ({name})'):
        docs = candidates.get(qid, [])[:3] if 'PRF' in name else []
        context = '\n'.join(passage_text[d] for d in docs)
        prompt = tmpl.format(query=queries[qid], docs=context)
        inputs = tokenizer(prompt, return_tensors='pt', truncation=True, padding=True).to(device)
        out = model.generate(**inputs, max_new_tokens=32)
        gen = tokenizer.decode(out[0], skip_special_tokens=True).strip()
        # for CoT_distilled pick last line
        if name == 'CoT_distilled': gen = gen.splitlines()[-1]
        expansions[qid] = gen


    # score expanded queries
    run_exp = {}
    for qid, exp in tqdm(expansions.items(), desc=f'Scoring ({name})'):
        docs = candidates.get(qid, [])
        if not docs: continue
        exp_q = queries[qid] + ' ' + exp
        pairs = [(exp_q, passage_text[d]) for d in docs]
        scores = cross_encoder.predict(pairs, batch_size=64).tolist()
        run_exp[qid] = {d: float(s) for d, s in zip(docs, scores)}


    # evaluate
    common = set(baseline_run) & set(run_exp)
    evaluator = pytrec_eval.RelevanceEvaluator(
        {qid: qrels[qid] for qid in common}, {'ndcg_cut.10','recall_100','map_cut_1000'}
    )
    metrics = evaluator.evaluate({qid: run_exp[qid] for qid in common})
    results[name] = {
        'NDCG@10': mean(m['ndcg_cut_10'] for m in metrics.values()),
        'Recall@100': mean(m['recall_100'] for m in metrics.values()),
        'MAP@1000': mean(m['map_cut_1000'] for m in metrics.values())
    }

```

```python
for name, mets in results.items():
    print(f"{name}: {mets}")
```
